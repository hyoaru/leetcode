### Contains Duplicate
Given an integer array nums, return true if any value appears more than once in the array, otherwise return false.

Example 1:
```
Input: nums = [1, 2, 3, 3]
Output: true
```

Example 2:
```
Input: nums = [1, 2, 3, 4]
Output: false
```

Constraints:
```
0 <= nums.length <= 10^5
-10^9 <= nums[i] <= 10^9
```

````python
class Solution:
    def hasDuplicate(self, nums: List[int]) -> bool:
        map = {}
        for num in nums:
            if map.get(num) == None:
                map[num] = 0
            else:
                map[num] += 1
                return map[num] > 0

        return False
        
Solution().hasDuplicate(nums=[1,2,3,3])
````
